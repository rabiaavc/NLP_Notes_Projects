## BAG OF WORDS NEDİR

📌 Bag of Words (BoW) Nedir?

Bag of Words (Kelime Torbası), bir metni sayısal forma dönüştürme yöntemidir.

Mantık çok basit:

Metindeki kelimeleri say → sırayı ve dil bilgisini önemseme → sadece kaç kez geçtiğine bak.

Yani:

Kelimelerin sırası önemli değil ❌

Anlam yapısı önemli değil ❌

Kaç kez geçtiği önemli ✅

⚠️ BoW’un Problemleri

-- Kelime sırasını kaybeder

-- Anlam ilişkisini kaybeder

-- Çok büyük kelime havuzu oluşabilir

-- Seyrek (sparse) matris oluşur
